## Enunciado resumido y contrato

En Morse, cada carácter se escribe con puntos (`.`) y rayas (`-`), y:

- **un espacio** separa los códigos de dos caracteres;
- **tres espacios** separan dos palabras;
- los espacios sobrantes del principio y del final no significan nada
  y hay que ignorarlos.

Hay que escribir `decode_morse(morse_code)`, que recibe la señal Morse
y devuelve el texto legible para una persona:

```python
decode_morse('.... . -.--   .--- ..- -.. .')
# debe devolver 'HEY JUDE'
```

| Recibe | Devuelve |
|---|---|
| `morse_code`: `str` ASCII con `.`, `-` y espacios | `str` en mayúsculas, con las palabras separadas por un espacio |

**Garantía del kata:** todas las entradas de los tests son Morse válido
con los separadores correctos, así que no hace falta comprobar errores.

**Nota honesta sobre los bordes:** si la entrada fuera una cadena vacía
(`''`) o solo espacios (`'   '`), esta versión —la de los dos bucles—
lanzaría `KeyError: ''`: después del `strip()` quedaría una «palabra»
vacía y `''` no es una clave de `MORSE_CODE`. El kata no exige ese
caso, así que aquí lo documentamos pero **no** lo tratamos ni lo
incluimos entre los tests válidos garantizados.


````python
# ---------------------------------------------------------------
# MORSE_CODE: la tabla de traducción del kata.
# En Codewars este diccionario ya viene precargado en el entorno de
# la kata; aquí lo definimos nosotros para que el cuaderno sea
# autocontenido. Clave: el código Morse de UN carácter.
# Valor: el carácter que significa.
# ---------------------------------------------------------------

MORSE_CODE = {                          # abre el diccionario: cada pareja es 'código': 'carácter'
    '.-': 'A', '-...': 'B', '-.-.': 'C', '-..': 'D', '.': 'E', '..-.': 'F',   # letras A-F
    '--.': 'G', '....': 'H', '..': 'I', '.---': 'J', '-.-': 'K', '.-..': 'L', # letras G-L
    '--': 'M', '-.': 'N', '---': 'O', '.--.': 'P', '--.-': 'Q', '.-.': 'R',   # letras M-R
    '...': 'S', '-': 'T', '..-': 'U', '...-': 'V', '.--': 'W', '-..-': 'X',   # letras S-X
    '-.--': 'Y', '--..': 'Z',                                                 # letras Y, Z
    '-----': '0', '.----': '1', '..---': '2', '...--': '3',                   # dígitos 0-3
    '....-': '4', '.....': '5', '-....': '6', '--...': '7',                   # dígitos 4-7
    '---..': '8', '----.': '9',                                               # dígitos 8, 9
    '.-.-.-': '.', '--..--': ',', '..--..': '?', ".----.": "'",               # signos . , ? '
    '-.-.--': '!', '-..-.': '/', '-.--.': '(', '-.--.-': ')',                 # signos ! / ( )
    '.-...': '&', '---...': ':', '-.-.-.': ';', '-...-': '=',                 # signos & : ; =
    '.-.-.': '+', '-....-': '-', '..--.-': '_', '.-..-.': '"',                # signos + - _ "
    '...-..-': '$', '.--.-.': '@',                                            # signos $ @
    '...---...': 'SOS',                                                       # código especial: SOS vale por un carácter
}                                       # cierra el diccionario

print(f"MORSE_CODE cargado: {len(MORSE_CODE)} códigos")   # confirma que la tabla está disponible
````

## La solución, línea a línea

La idea que ordena todo: **primero se corta lo grande (las palabras) y
luego lo pequeño (las letras)**. Si cortáramos por un espacio el
mensaje entero, los huecos de 3 espacios dejarían cadenas vacías `''`
que el diccionario no reconoce.


````python
def decode_morse(morse_code):                        # define la función; 'morse_code' es la señal Morse recibida (str)
    palabras = morse_code.strip().split('   ')       # strip() quita los espacios de los bordes; split('   ') corta por cada hueco de 3 espacios -> lista de palabras en Morse
    resultado = []                                   # lista vacía: aquí guardaremos cada palabra ya traducida a texto
    for palabra in palabras:                         # bucle exterior: recorre la lista palabra a palabra
        texto = ''                                   # cadena vacía donde se acumulan las letras de la palabra actual
        for codigo in palabra.split(' '):            # bucle interior: corta la palabra por 1 espacio y recorre sus códigos de letra
            texto += MORSE_CODE[codigo]              # consulta el código en el diccionario y añade su carácter al final de 'texto'
        resultado.append(texto)                      # la palabra ya está completa: la guardamos en la lista 'resultado'
    return ' '.join(resultado)                       # une las palabras con un espacio y devuelve el mensaje descifrado
````

## Cómo evolucionan las variables (ejemplo `'HEY JUDE'`)

Con la entrada `'  .... . -.--   .--- ..- -.. .  '` (lleva espacios de
más en los bordes, a propósito):

| Paso | Variable | Valor |
|---|---|---|
| `strip()` + `split('   ')` | `palabras` | `['.... . -.--', '.--- ..- -.. .']` |
| vuelta 1 del bucle exterior | `palabra` | `'.... . -.--'` |
| bucle interior | `codigo` → `texto` | `'....'`→`'H'`, `'.'`→`'HE'`, `'-.--'`→`'HEY'` |
| `append` | `resultado` | `['HEY']` |
| vuelta 2 del bucle exterior | `palabra` | `'.--- ..- -.. .'` |
| bucle interior | `codigo` → `texto` | `'.---'`→`'J'`, `'..-'`→`'JU'`, `'-..'`→`'JUD'`, `'.'`→`'JUDE'` |
| `append` | `resultado` | `['HEY', 'JUDE']` |
| `return ' '.join(...)` | — | `'HEY JUDE'` |

La celda siguiente repite la misma solución con `print` dentro para
ver esa evolución en vivo.


````python
def decode_morse_traza(morse_code):                      # misma solución, con prints para observar las variables
    palabras = morse_code.strip().split('   ')           # limpia los bordes y corta el mensaje en palabras
    print('palabras =', palabras)                        # muestra la lista de palabras en Morse
    resultado = []                                       # lista vacía para las palabras traducidas
    for palabra in palabras:                             # recorre cada palabra
        texto = ''                                       # reinicia la traducción de la palabra actual
        for codigo in palabra.split(' '):                # recorre cada código de letra de la palabra
            texto += MORSE_CODE[codigo]                  # traduce el código y lo acumula
            print(f'  {codigo!r:12} -> {MORSE_CODE[codigo]!r:5} | texto = {texto!r}')  # muestra cada traducción parcial
        resultado.append(texto)                          # guarda la palabra ya completa
        print('resultado =', resultado)                  # muestra la lista tras añadir la palabra
    return ' '.join(resultado)                           # une con espacios y devuelve

print('devuelve:', repr(decode_morse_traza('  .... . -.--   .--- ..- -.. .  ')))  # ejecuta la traza con el ejemplo
````

## Tests autocontenidos (10)

Los mismos casos del cuaderno de ejercicio: el ejemplo del enunciado,
los dos tipos de separador, los bordes, SOS, dígitos y signos.


````python
_aciertos = 0                                            # contador de tests superados
_fallos = 0                                              # contador de tests fallados

def probar(entrada, esperado):                           # ejecuta decode_morse con 'entrada' y compara con 'esperado'
    global _aciertos, _fallos                            # permite modificar los contadores definidos fuera
    try:                                                 # intenta ejecutar la solución
        obtenido = decode_morse(entrada)                 # llama a la función con la entrada del test
    except Exception as e:                               # si lanza cualquier excepción, cuenta como fallo
        obtenido = f'ERROR {type(e).__name__}: {e}'      # guarda el error como "valor obtenido"
    if obtenido == esperado:                             # ¿coincide lo obtenido con lo esperado?
        _aciertos += 1                                   # sí: suma un acierto
        print(f'    OK      {entrada!r} -> {obtenido!r}')          # informa del acierto
    else:                                                # no coincide
        _fallos += 1                                     # suma un fallo
        print(f'    FALLO   {entrada!r} -> obtenido {obtenido!r}, esperado {esperado!r}')  # informa del fallo

print('== Ejemplo del enunciado ==')                     # primer bloque de casos
probar('.... . -.--   .--- ..- -.. .', 'HEY JUDE')       # el caso del enunciado

print('== Un espacio separa letras, tres separan palabras ==')  # segundo bloque
probar('.- -...', 'AB')                                  # dos letras juntas en una palabra
probar('.-   -...', 'A B')                               # dos palabras de una letra

print('== Bordes ==')                                    # tercer bloque
probar('.-', 'A')                                        # una sola letra
probar('   .... . -.--   ', 'HEY')                       # espacios sobrantes delante y detrás
probar('.... . -.--   ', 'HEY')                          # espacios sobrantes solo al final

print('== SOS, dígitos y signos ==')                     # cuarto bloque
probar('...---...', 'SOS')                               # SOS es un único código, no tres letras
probar('...---... -.-.--', 'SOS!')                       # SOS seguido de un signo
probar('.... . .-.. .--.   ...---...', 'HELP SOS')       # SOS como palabra suelta
probar('.---- ..--- ...--', '123')                       # dígitos

_total = _aciertos + _fallos                             # total de casos ejecutados
print(f'Resultado: {_aciertos}/{_total} tests superados')          # marcador final
````
