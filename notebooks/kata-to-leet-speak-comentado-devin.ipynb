## Enunciado resumido y contrato

Una función `to_leet_speak(texto)` que traduce a «leet» (1337) una
frase escrita en MAYÚSCULAS con espacios. Cada letra se cambia por su
símbolo según el alfabeto del kata y **los espacios se conservan**.

| Recibe | Devuelve |
|---|---|
| `texto`: `str` con solo letras mayúsculas inglesas y espacios | `str` con cada letra sustituida y los espacios intactos |

Ejemplos: `'LEET'` → `'1337'` · `'I LOVE PYTHON'` → `'! 10V3 PY7#0N'`
· `''` → `''` (la cadena vacía está aceptada por el contrato).

**Aviso:** el leet es una **codificación didáctica, no un cifrado
seguro** — el mapeo es público y no usa clave, así que no protege
información. Sirve para practicar diccionarios y bucles.


```python
LEET = {                                        # diccionario fijo del kata: cada letra -> su símbolo leet
    'A': '@', 'B': '8', 'C': '(', 'D': 'D', 'E': '3', 'F': 'F',   # A-F (D y F se quedan igual)
    'G': '6', 'H': '#', 'I': '!', 'J': 'J', 'K': 'K', 'L': '1',   # G-L (J y K se quedan igual)
    'M': 'M', 'N': 'N', 'O': '0', 'P': 'P', 'Q': 'Q', 'R': 'R',   # M-R (estas cinco se quedan igual)
    'S': '$', 'T': '7', 'U': 'U', 'V': 'V', 'W': 'W', 'X': 'X',   # S-X (U, V, W, X se quedan igual)
    'Y': 'Y', 'Z': '2',                                         # Y, Z
}                                               # el espacio NO está en el diccionario: esa es la pista
```

## La solución, línea a línea

Un diccionario y un solo bucle: recorremos la entrada carácter a
carácter; si la letra está en `LEET` añadimos su símbolo y si no está
(los espacios) la copiamos tal cual.


```python
def to_leet_speak(texto):                          # define la función; 'texto' es la frase en mayúsculas
    resultado = ''                                 # cadena vacía donde acumulamos la traducción
    for letra in texto:                            # bucle: recorre la entrada un carácter cada vuelta
        if letra in LEET:                          # ¿existe la letra como clave del diccionario?
            resultado = resultado + LEET[letra]    # sí: añade su símbolo leet al final del resultado
        else:                                      # no está en el diccionario (los espacios no están)
            resultado = resultado + letra          # copia el carácter tal cual: así se conservan los espacios
    return resultado                               # devuelve la traducción completa
```

## Cómo evolucionan las variables (ejemplo `'LEET'`)

| Vuelta | `letra` | ¿`letra in LEET`? | `resultado` |
|---|---|---|---|
| 1 | `'L'` | sí → `'1'` | `'1'` |
| 2 | `'E'` | sí → `'3'` | `'13'` |
| 3 | `'E'` | sí → `'3'` | `'133'` |
| 4 | `'T'` | sí → `'7'` | `'1337'` |

Y con un espacio de por medio (`'A B'`): `'A'`→`'@'`, `' '`→`' '`
(no está en `LEET`, se copia tal cual), `'B'`→`'8'`: sale `'@ 8'`.

La celda siguiente repite la misma solución con `print` dentro para
ver esa evolución en vivo.


```python
def to_leet_speak_traza(texto):                    # misma solución, con prints para observar las variables
    resultado = ''                                 # cadena vacía donde acumulamos la traducción
    for letra in texto:                            # recorre la entrada carácter a carácter
        if letra in LEET:                          # ¿tiene la letra sustituto en el diccionario?
            resultado = resultado + LEET[letra]    # sí: añade el símbolo leet
            print(f'  {letra!r:5} esta en LEET -> {LEET[letra]!r:5} | resultado = {resultado!r}')  # muestra la sustitución
        else:                                      # no está (los espacios)
            resultado = resultado + letra          # copia el carácter tal cual
            print(f'  {letra!r:5} NO esta        -> {letra!r:5} | resultado = {resultado!r}')    # muestra la copia literal
    return resultado                               # devuelve la traducción completa

print('devuelve:', repr(to_leet_speak_traza('A B')))   # traza con un espacio de por medio
print()                                              # línea en blanco para separar las dos trazas
print('devuelve:', repr(to_leet_speak_traza('LEET')))  # traza con el ejemplo del enunciado
```

## Tests autocontenidos (10)

Los mismos 10 casos del cuaderno de ejercicio, incluida la cadena
vacía (que el contrato sí acepta) y las letras que se traducen a sí
mismas.


```python
CASOS = [                                            # lista de tuplas (entrada, salida esperada)
    ('LEET', '1337'),                                # ejemplo del enunciado
    ('', ''),                                        # cadena vacía: aceptada por el contrato
    ('HOLA MUNDO', '#01@ MUND0'),                    # frase con un espacio
    ('CODEWARS', '(0D3W@R$'),                        # nombre de la plataforma
    ('THE QUICK BROWN FOX', '7#3 QU!(K 8R0WN F0X'),  # pangrama con varios espacios
    ('A  B', '@  8'),                                # dos espacios seguidos se conservan
    ('DFJKMNPQRUVWXY', 'DFJKMNPQRUVWXY'),            # letras que se traducen a sí mismas
    ('SOS', '$0$'),                                  # tres letras con sustituto
    ('ABCDEFGHIJKLMNOPQRSTUVWXYZ', '@8(D3F6#!JK1MN0PQR$7UVWXY2'),  # el alfabeto entero
    ('I LOVE PYTHON', '! 10V3 PY7#0N'),              # ejemplo final del enunciado
]

def probar():                                        # ejecuta los 10 casos y resume
    aciertos = 0                                     # contador de tests superados
    for i, (entrada, esperada) in enumerate(CASOS, 1):   # recorre los casos numerándolos desde 1
        try:                                         # intenta ejecutar la solución
            obtenida = to_leet_speak(entrada)        # llama a la función con la entrada del test
        except Exception as exc:                     # captura cualquier error inesperado
            obtenida = 'ERROR: %r' % exc             # lo guarda como "valor obtenido"
        if obtenida == esperada:                     # compara lo obtenido con lo esperado
            aciertos += 1                            # acierto
            print('test %02d  OK      %r -> %r' % (i, entrada, esperada))          # informa del acierto
        else:                                        # fallo
            print('test %02d  FALLO   %r -> esperaba %r, obtuvo %r'                # informa del fallo
                  % (i, entrada, esperada, obtenida))
    print('Resumen: %d/10 aprobados' % aciertos)     # marcador final

probar()                                             # lanza la batería de tests
```
