El «leet» o «1337» es una escritura juguetona que cambia letras por
números y símbolos parecidos. Aquí es una excusa para practicar
diccionarios y bucles. Ojo: es una **codificación didáctica**, no un
cifrado seguro.



## Qué vas a construir

Una función `to_leet_speak(texto)` que recibe una frase en MAYÚSCULAS
con espacios y devuelve la misma frase traducida a leet.

- `to_leet_speak('LEET')` → `'1337'`
- `to_leet_speak('I LOVE PYTHON')` → `'! 10V3 PY7#0N'`
- Los espacios se quedan como espacios.
- Solo llegan letras mayúsculas inglesas y espacios: no hay que
  preocuparse por minúsculas ni números.

## Pistas graduadas

Lee solo hasta donde necesites; cada pista destapa un poco más.

1. El resultado se construye poco a poco. Empieza con una cadena
   vacía: `resultado = ''`.
2. Para cada carácter `letra` de `texto`, la expresión
   `letra in LEET` te dice si tiene sustituto.
3. Para añadir algo al final de una cadena:
   `resultado = resultado + algo`.
4. En palabras: recorre `texto` con un `for`; si la letra está en
   `LEET`, añade `LEET[letra]`; si no está (como los espacios),
   añade la letra misma. Al terminar el bucle, devuelve `resultado`.

````python
CASOS = [
    ('LEET', '1337'),
    ('', ''),
    ('HOLA MUNDO', '#01@ MUND0'),
    ('CODEWARS', '(0D3W@R$'),
    ('THE QUICK BROWN FOX', '7#3 QU!(K 8R0WN F0X'),
    ('A  B', '@  8'),
    ('DFJKMNPQRUVWXY', 'DFJKMNPQRUVWXY'),
    ('SOS', '$0$'),
    ('ABCDEFGHIJKLMNOPQRSTUVWXYZ', '@8(D3F6#!JK1MN0PQR$7UVWXY2'),
    ('I LOVE PYTHON', '! 10V3 PY7#0N'),
]


def probar():
    aciertos = 0
    for i, (entrada, esperada) in enumerate(CASOS, 1):
        try:
            obtenida = to_leet_speak(entrada)
        except Exception as exc:
            obtenida = 'ERROR: %r' % exc
        if obtenida == esperada:
            aciertos += 1
            print('test %02d  OK      %r -> %r' % (i, entrada, esperada))
        else:
            print('test %02d  FALLO   %r -> esperaba %r, obtuvo %r'
                  % (i, entrada, esperada, obtenida))
    print('Resumen: %d/10 aprobados' % aciertos)


probar()
````
