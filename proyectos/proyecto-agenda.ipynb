### Versión de referencia (comentada)

```python
def agregar_contacto(agenda, nombre, telefono):
    agenda[nombre] = telefono      # clave → valor
    return True

def ver_contactos(agenda):
    if not agenda:                 # dict vacío es "falso"
        print("La agenda está vacía.")
        return                     # return a secas: sale devolviendo None
    for nombre, tel in agenda.items():
        print(f"{nombre}: {tel}")

def buscar_contacto(agenda, nombre):
    return agenda.get(nombre)      # get devuelve None si no existe

def eliminar_contacto(agenda, nombre):
    if nombre in agenda:           # comprobar antes de borrar
        del agenda[nombre]
        return True
    return False                   # no existía
```

**Fíjate:** `main()` no devuelve nada — es una función *de efecto*. Las otras sí: devuelven datos o `True/False`, que es justo lo que los tests comprueban.
